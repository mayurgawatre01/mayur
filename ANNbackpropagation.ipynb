```python

import numpy as np

# Input data (XOR problem)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

# Expected output
y = np.array([[0],
              [1],
              [1],
              [0]])

# Initialize weights and biases
np.random.seed(42)

w1 = np.random.rand(2, 3)
w2 = np.random.rand(3, 1)

b1 = np.zeros((1, 3))
b2 = np.zeros((1, 1))

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Training using Backpropagation
for epoch in range(10000):

    # Forward propagation
    hidden = sigmoid(np.dot(X, w1) + b1)
    output = sigmoid(np.dot(hidden, w2) + b2)

    # Calculate error
    error = y - output

    # Backward propagation
    d_output = error * output * (1 - output)

    d_hidden = np.dot(d_output, w2.T) * hidden * (1 - hidden)

    # Update weights
    w2 += np.dot(hidden.T, d_output) * 0.5
    w1 += np.dot(X.T, d_hidden) * 0.5

    # Update biases
    b2 += np.sum(d_output, axis=0, keepdims=True) * 0.5
    b1 += np.sum(d_hidden, axis=0, keepdims=True) * 0.5

# Display results
print("Predicted Output:")
print(output)

print("\nActual Output:")
print(y)
```
