```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np

# Step 2: Load Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Step 3: Display first five records
print("Original Dataset:")
print(df.head())

# Step 4: Check dataset information
print("\nDataset Information:")
print(df.info())

# Step 5: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 6: Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

# Step 7: Remove unnecessary columns
df.drop(['Cabin', 'Name', 'Ticket', 'PassengerId'],
        axis=1, inplace=True)

# Step 8: Remove duplicate records
df.drop_duplicates(inplace=True)

# Step 9: Convert categorical data into numerical data
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
df = pd.get_dummies(df, columns=['Embarked'], dtype=int)

# Step 10: Display cleaned dataset
print("\nPreprocessed Dataset:")
print(df.head())

# Step 11: Check missing values after preprocessing
print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())

print("\nDataset Shape:", df.shape)
 make only pdf of this code
```
