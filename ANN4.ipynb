```python

import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 100)

# 1. Binary Step Function
def binary_step(x):
    return np.where(x >= 0, 1, 0)

# 2. Linear Activation Function
def linear(x):
    return x

# 3. Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# 4. Tanh Activation Function
def tanh(x):
    return np.tanh(x)

# 5. ReLU Activation Function
def relu(x):
    return np.maximum(0, x)

# 6. Leaky ReLU Activation Function
def leaky_relu(x):
    return np.maximum(0.01 * x, x)

# Plot all activation functions
plt.figure(figsize=(12, 8))

# Binary Step
plt.subplot(2, 3, 1)
plt.plot(x, binary_step(x))
plt.title("Binary Step")
plt.grid()

# Linear
plt.subplot(2, 3, 2)
plt.plot(x, linear(x))
plt.title("Linear")
plt.grid()

# Sigmoid
plt.subplot(2, 3, 3)
plt.plot(x, sigmoid(x))
plt.title("Sigmoid")
plt.grid()

# Tanh
plt.subplot(2, 3, 4)
plt.plot(x, tanh(x))
plt.title("Tanh")
plt.grid()

# ReLU
plt.subplot(2, 3, 5)
plt.plot(x, relu(x))
plt.title("ReLU")
plt.grid()

# Leaky ReLU
plt.subplot(2, 3, 6)
plt.plot(x, leaky_relu(x))
plt.title("Leaky ReLU")
plt.grid()

plt.tight_layout()
plt.show()

print("All activation functions implemented successfully!")
```
